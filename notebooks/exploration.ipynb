## 01.1-VARIABLES CONSTANTES, TYPES RÉELS ET VARIABLES QUASI BINAIRES

Il se peut que notre dataset ait été prénettoyé et/ou compressé (l'existence de valeurs "suspectes" nous pousse à considérer ce scénario)

**CONCLUSION :**
- `distinct_user_agents` est constante : à exclure.
- Cast de 8 colonnes float ne contiennent que des entiers : à convertir en `int` dans le dataset clean.
- `auth_failure_rate` et `distinct_usernames` ne prennent que 0/1 : ce sont des indicateurs binaires malgré le nom "rate".
- `max_depth` et `distinct_depths` sont discrètes (catégories ordonnées).

```python
# 1. Variables constantes (variance nulle : aucune information)
colonnes_constantes = [c for c in df.columns if df[c].nunique(dropna=False) <= 1]
print("Colonnes constantes :", colonnes_constantes)

# 2. Colonnes float qui ne contiennent que des entiers
colonnes_float = df.select_dtypes("float").columns
float_entiers = [c for c in colonnes_float if (df[c] == df[c].round()).all()]
print("\nFloat contenant uniquement des entiers (à convertir en int) :")
print(float_entiers)

# 3. Variables quasi binaires (2 valeurs seulement)
quasi_binaires = [c for c in df.columns if df[c].nunique() == 2]
print("\nVariables à 2 valeurs :")
for c in quasi_binaires:
    print(f"  {c:20s} -> {sorted(df[c].unique().tolist())}")

# 4. Variables discrètes à faible cardinalité (3 à 5 valeurs)
discretes = [c for c in df.select_dtypes("number").columns if 2 < df[c].nunique() <= 5]
print("\nVariables discrètes (3 à 5 valeurs) :")
for c in discretes:
    print(f"  {c:20s} -> {sorted(df[c].unique().tolist())}")
display(df.groupby("profile")["auth_failure_count"].describe().round(2))
```

## 01.2-DÉPENDANCES FONCTIONNELLES (Determine == "→")

X détermine Y si chaque valeur de X correspond à une seule valeur de Y.


**CONCLUSION :**

ON A : (Determine == "→")
- `profile → label`
- `profile → auth_failure_rate`
- `auth_failure_count → auth_failure_rate`

Les sens inverses sont faux (ex. `auth_failure_rate = 0` correspond à 7 profils). `profile` ne doit pas être utilisé comme variable explicative du modèle 1 (il détermine la cible) ; la hiérarchie bot / profil est cohérente.

```python
def determine(data, x, y):
    """True si x -> y (aucune valeur de x associée à plusieurs valeurs de y)."""
    return bool((data.groupby(x)[y].nunique(dropna=False) <= 1).all())


for x, y in [("profile", "label"), ("profile", "auth_failure_rate")]:
    print(f"{x} -> {y} : {determine(df, x, y)}   |   {y} -> {x} : {determine(df, y, x)}")

print()
print(pd.crosstab(df["profile"], df["label"]))
print()
print(pd.crosstab(df["profile"], df["auth_failure_rate"]))

# Scan automatique sur les colonnes à faible cardinalité (<= 50 valeurs)
candidats = [c for c in df.columns if 1 < df[c].nunique() <= 50 and c not in colonnes_constantes]
dependances = [(x, y) for x in candidats for y in candidats if x != y and determine(df, x, y)]
print("\nScan automatique :")
for x, y in dependances:
    print(f"  {x} -> {y}")
```

## 01.3-REDONDANCES

! Seuil de correlation = 0.95 !

**CONCLUSION :**

ON A REDONDANCE : (similrité et calculable)
- `auth_failure_rate` et `distinct_usernames` sont identiques
- `auth_failure_count = auth_failure_rate × request_count`
- `path_repeat_ratio = 1 − distinct_paths / request_count`

```python
colonnes_num = [c for c in df.select_dtypes("number").columns if c not in colonnes_constantes]

# 1. Colonnes strictement identiques
print("Colonnes identiques :")
trouve = False
for i, a in enumerate(colonnes_num):
    for b in colonnes_num[i + 1:]:
        if df[a].equals(df[b]):
            print(f"  {a} == {b}")
            trouve = True
if not trouve:
    print("  aucune")

# 2. Colonnes obtenues par formule
formules = {
    "auth_failure_count = auth_failure_rate x request_count":
        (df["auth_failure_count"], df["auth_failure_rate"] * df["request_count"]),
    "path_repeat_ratio = 1 - distinct_paths / request_count":
        (df["path_repeat_ratio"], 1 - df["distinct_paths"] / df["request_count"]),
}
print("\nRelations par formule :")
for nom, (reel, calcule) in formules.items():
    ecart = float(np.abs(reel - calcule).max())
    print(f"  {nom:58s} écart max = {ecart:.2e} -> {'OUI' if ecart < 1e-9 else 'NON'}")

# 3. Paires fortement corrélées (|r| >= 0.95)
corr = df[colonnes_num].corr().abs()
haut = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool)).stack()
print("\nPaires avec |corrélation| >= 0.95 :")
print(haut[haut >= 0.95].sort_values(ascending=False).round(3).to_string())
```

## 01.4-COHÉRENCE LOGIQUE ET BORNES (VALIDE)

Vérifie que les valeurs respectent les contraintes que doit satisfaire un vrai log de sessions.

**CONCLUSION :**

- Toutes les contraintes logiques sont respectées : pas d'incohérence à corriger.
- `path_entropy` contient des `-0.0` (zéros négatifs) : cosmétique, normalisable avec `+ 0.0`.
- Les lignes `max_depth = 0` (sessions d'une seule page) sont valides : à conserver.

```python
taux = ["path_repeat_ratio", "static_asset_ratio", "error_rate", "not_found_rate", "auth_failure_rate"]
eps = 1e-9

verifications = {
    "taux hors [0, 1]": ((df[taux] < 0) | (df[taux] > 1)).any(axis=1).sum(),
    "request_count <= 0": (df["request_count"] <= 0).sum(),
    "requests_per_minute <= 0": (df["requests_per_minute"] <= 0).sum(),
    "interval_cv < 0": (df["interval_cv"] < 0).sum(),
    "distinct_paths > request_count": (df["distinct_paths"] > df["request_count"]).sum(),
    "auth_failure_count > request_count": (df["auth_failure_count"] > df["request_count"]).sum(),
    "error_rate < auth_failure_rate": (df["error_rate"] < df["auth_failure_rate"] - eps).sum(),
    "error_rate < not_found_rate": (df["error_rate"] < df["not_found_rate"] - eps).sum(),
    "error_rate < auth_failure_rate + not_found_rate": (df["error_rate"] < df["auth_failure_rate"] + df["not_found_rate"] - eps).sum(),
    "path_entropy < 0": (df["path_entropy"] < 0).sum(),
    "path_entropy > log2(distinct_paths)": (df["path_entropy"] > np.log2(df["distinct_paths"]) + eps).sum(),
    "distinct_paths == 1 et entropy != 0": ((df["distinct_paths"] == 1) & (df["path_entropy"].abs() > eps)).sum(),
    "distinct_depths > max_depth + 1": (df["distinct_depths"] > df["max_depth"] + 1).sum(),
}
resultat = pd.Series(verifications, name="violations").to_frame()
resultat["statut"] = np.where(resultat["violations"] == 0, "OK", "ANOMALIE")
print(resultat)

# Détail cosmétique : zéros négatifs (-0.0)
zeros_negatifs = ((df["path_entropy"] == 0) & np.signbit(df["path_entropy"])).sum()
print(f"\npath_entropy = -0.0 : {zeros_negatifs} lignes")

# Cas limites : sessions sans profondeur
print("\nLignes avec max_depth = 0 :")
print(df[df["max_depth"] == 0].T)
```

## 01.5-DISTRIBUTIONS PAR PROFIL, BORNES ARTIFICIELLES ET OUTLIERS

**CONCLUSION :**
- Les bornes sont tronquées net par profil (ex. `api_client` ≤ 13 requêtes, `headless_browser` ≥ 52, `scraper` ≤ 49) 
- Les outliers globaux sont des profils légitimes (ex. `headless_browser` sur `request_count`) : a garder . Toute analyse d'outliers se fait par profil.
- `static_asset_ratio` vaut toujours 0 pour 4 profils.
- Aucun doublon ni conflit sur les features.

```python
cles = ["request_count", "requests_per_minute", "interval_cv", "distinct_paths", "path_entropy"]

# 1. Min / médiane / max par profil (bornes tronquées = signe de données générées)
print(df.groupby("profile")[cles].agg(["min", "median", "max"]).round(2))

# 2. static_asset_ratio : plage par profil
print("\nstatic_asset_ratio par profil :")
print(df.groupby("profile")["static_asset_ratio"].agg(["min", "max"]).round(3))


# 3. Outliers (règle IQR) : global vs par profil
def taux_outliers(s):
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    return ((s < q1 - 1.5 * iqr) | (s > q3 + 1.5 * iqr)).mean() * 100


tableau = pd.DataFrame({
    "outliers_global_%": df[cles].apply(taux_outliers),
    "outliers_moyenne_par_profil_%": df.groupby("profile")[cles].agg(taux_outliers).mean(),
}).round(1)
print("\nOutliers IQR :")
print(tableau)

# 4. Doublons et conflits sur les features seules
features = [c for c in df.columns if c not in ["label", "profile"] + colonnes_constantes]
print("\nDoublons sur les features :", int(df.duplicated(subset=features).sum()))
conflits = (df.groupby(features)["profile"].nunique() > 1).sum()
print("Mêmes features avec profils différents :", int(conflits))
```

## 01.6-ASYMÉTRIE ET ÉCHELLES

**CONCLUSION :**

- `request_count`, `distinct_paths` et `requests_per_minute` sont asymétriques et non bornées :choix de  `log1p` utile pour les modèles linéaires. Les taux, binaires et discrètes ne sont pas concernés.
- Les échelles sont très hétérogènes (taux dans [0, 1], `requests_per_minute` jusqu'à ~785) : standardisation nécessaire pour les modèles sensibles à l'échelle, ajustée **après** le split, sur le train seulement.

```python
num = df[[c for c in df.select_dtypes("number").columns if c not in colonnes_constantes + ["label"]]]
stats = pd.DataFrame({
    "min": num.min(),
    "max": num.max(),
    "std": num.std(),
    "skew": num.skew(),
})
# log1p : uniquement variables continues non bornées (pas de taux, pas de binaires/discrètes)
stats["log1p_candidate"] = (
    (stats["skew"].abs() > 1) & (stats["min"] >= 0) & (stats["max"] > 1) & (num.nunique() > 20)
)
print(stats.round(3))

print("\nCandidates à log1p (continues, non bornées, |skew| > 1) :", stats.index[stats["log1p_candidate"]].tolist())
print("Rapport d'échelle max/min des écarts-types :", round(stats["std"].max() / stats["std"].min(), 1))
```

# 02-GÉNÉRATION DU DATASET CLEAN (TRAIN / TEST)

Applique les décisions de 01.8 :

1. Suppression des colonnes constantes et redondantes.
2. Conversion float → int des colonnes qui ne contiennent que des entiers.
3. Normalisation des zéros négatifs de `path_entropy` (`-0.0` → `0.0`).
4. Mélange du dataset (trié par profil) puis split **stratifié sur `profile`** (80 % / 20 %).
5. Construction des jeux d'entraînement des deux modèles :
   - **Modèle 1** (bot / non-bot) : toutes les lignes, cible `label`.
   - **Modèle 2** (profil du bot) : lignes `label == 1` uniquement, cible `profile`.
6. Contrôles de qualité (pas de manquants, pas de constantes, pas de fuite train/test, stratification respectée).

> Le `log1p` et la standardisation ne sont **pas** appliqués ici : ils seront placés dans un `Pipeline` pour être ajustés sur le train seulement, ce qui évite toute fuite de données.

```python
from sklearn.model_selection import train_test_split

SEED = 42
TEST_SIZE = 0.2

# 1. Suppression des colonnes constantes et redondantes (décisions de 01.8)
colonnes_a_supprimer = colonnes_constantes + redondantes_a_supprimer
df_clean = df.drop(columns=colonnes_a_supprimer).copy()

# 2. Conversion float -> int (colonnes ne contenant que des entiers)
cols_int = [c for c in float_entiers if c in df_clean.columns]
df_clean[cols_int] = df_clean[cols_int].astype(int)

# 3. Normalisation des zéros négatifs (-0.0 -> 0.0)
df_clean["path_entropy"] = df_clean["path_entropy"] + 0.0

# 4. Split train / test : mélange + stratification sur le profil
df_train, df_test = train_test_split(
    df_clean,
    test_size=TEST_SIZE,
    shuffle=True,
    stratify=df_clean["profile"],
    random_state=SEED,
)

# 5. Jeux d'entraînement des deux modèles
features = [c for c in df_clean.columns if c not in ["label", "profile"]]

# Modèle 1 : bot / non-bot (toutes les lignes)
X_train_m1, y_train_m1 = df_train[features], df_train["label"]
X_test_m1, y_test_m1 = df_test[features], df_test["label"]

# Modèle 2 : profil du bot (lignes label == 1 uniquement)
bots_train = df_train[df_train["label"] == 1]
bots_test = df_test[df_test["label"] == 1]
X_train_m2, y_train_m2 = bots_train[features], bots_train["profile"]
X_test_m2, y_test_m2 = bots_test[features], bots_test["profile"]

# 6. Contrôles de qualité
assert df_clean.isna().sum().sum() == 0, "valeurs manquantes"
assert not [c for c in df_clean.columns if df_clean[c].nunique() <= 1], "colonne constante"
assert not (set(df_train.index) & set(df_test.index)), "fuite : lignes communes train/test"
assert len(df_train) + len(df_test) == len(df_clean), "lignes perdues"
assert not ((df_clean["path_entropy"] == 0) & np.signbit(df_clean["path_entropy"])).any(), "-0.0 restant"

prop_profil = pd.DataFrame({
    "train_%": df_train["profile"].value_counts(normalize=True) * 100,
    "test_%": df_test["profile"].value_counts(normalize=True) * 100,
}).round(2)
assert (prop_profil["train_%"] - prop_profil["test_%"]).abs().max() < 0.1, "stratification imparfaite"

corr = df_clean[features].corr().abs()
max_corr = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool)).stack().max()

print("df_clean :", df_clean.shape, "| colonnes supprimées :", colonnes_a_supprimer)
print("df_train :", df_train.shape, "| df_test :", df_test.shape)
print("\nTypes :\n", df_clean.dtypes.value_counts().to_string())
print(f"\nCorrélation max entre features : {max_corr:.3f}")
print("\nRépartition des profils :\n", prop_profil)
print("\nRépartition du label (%) :")
print(pd.DataFrame({"train": df_train["label"].value_counts(normalize=True) * 100,
                    "test": df_test["label"].value_counts(normalize=True) * 100}).round(2))
print("\nModèle 1 :", X_train_m1.shape, X_test_m1.shape)
print("Modèle 2 :", X_train_m2.shape, X_test_m2.shape, "| classes :", sorted(y_train_m2.unique()))
```

# 03.1 Stockage physique :

```python
import os

DOSSIER = "dataClean"                      

cols_phase1 = features + ["label"]
cols_phase2 = features + ["profile"]

jeux = {
    "modele1": (df_train[cols_phase1],
                df_test[cols_phase1]),
    "modele2": (df_train.loc[df_train["label"] == 1, cols_phase2],
                df_test.loc[df_test["label"] == 1, cols_phase2]),
}

for nom, (train, test) in jeux.items():
    os.makedirs(f"{DOSSIER}/{nom}", exist_ok=True)
    train.to_csv(f"{DOSSIER}/{nom}/train.csv", index=False)
    test.to_csv(f"{DOSSIER}/{nom}/test.csv", index=False)
    print(f"{nom} : train {train.shape} | test {test.shape}")

print("\nVérification :")
for nom in jeux:
    for partie in ("train", "test"):
        d = pd.read_csv(f"{DOSSIER}/{nom}/{partie}.csv")
        cible = d.columns[-1]
        assert d.isna().sum().sum() == 0, f"valeurs manquantes dans {nom}/{partie}"
        print(f"  {nom}/{partie}.csv : {d.shape[0]} lignes, {d.shape[1]} colonnes, cible = '{cible}', "
              f"classes = {sorted(d[cible].unique())}")
```
